# Eksplorasi Pertemuan 1 — Benchmark Manual vs NumPy
Ukur waktu perkalian matriks manual vs NumPy untuk berbagai ukuran.

```python
import sys; sys.path.insert(0, "../src")
import time
import numpy as np
from main import kali_matriks_manual, kali_matriks_numpy
```

```python
rng = np.random.default_rng(42)
ukuran, t_manual, t_numpy = [10, 50, 100], [], []
for n in ukuran:
    A, B = rng.random((n, n)), rng.random((n, n))
    t0 = time.perf_counter(); kali_matriks_manual(A, B); t_manual.append(time.perf_counter() - t0)
    t0 = time.perf_counter(); kali_matriks_numpy(A, B); t_numpy.append(time.perf_counter() - t0)
for n, tm, tn in zip(ukuran, t_manual, t_numpy):
    print(f"{n}x{n}: manual={tm*1000:.1f}ms numpy={tn*1000:.3f}ms speedup={tm/tn:.0f}x")
```

```python
# Tugas 1: plot kurva sigmoid
import matplotlib.pyplot as plt
x = np.linspace(-10, 10, 200)
# TODO: definisikan sigmoid lalu plt.plot(x, sigmoid(x))
def sigmoid(x):
    return 1 / (1 + np.exp(-x))

plt.plot(x, sigmoid(x))
plt.xlabel("x")
plt.ylabel("sigmoid(x)")
plt.title("Kurva Sigmoid")
plt.grid(True)
plt.show()
```
