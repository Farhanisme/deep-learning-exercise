```python
# Bagian 2 — model ekuivalen di Keras (butuh tensorflow)
import numpy as np
from tensorflow import keras

X = np.array([[0,0],[0,1],[1,0],[1,1]], dtype=float)
y = np.array([0.,1.,1.,0.])

model = keras.Sequential([
    keras.layers.Dense(2, activation="sigmoid", input_shape=(2,)),
    keras.layers.Dense(1, activation="sigmoid"),
])
model.compile(optimizer=keras.optimizers.SGD(learning_rate=0.5), loss="mse")
model.fit(X, y, epochs=5000, verbose=0)
print(model.predict(X, verbose=0).round(3))
```
