# Eksplorasi Pertemuan 3 — Decision Boundary Perceptron vs MLP
Visualisasikan wilayah keputusan pada bidang 2D.

```python
import sys; sys.path.insert(0, "../src")
import numpy as np
import matplotlib.pyplot as plt
from perceptron import Perceptron, forward_mlp, XOR_W1, XOR_B1, XOR_W2, XOR_B2
X = np.array([[0,0],[0,1],[1,0],[1,1]], dtype=float)
```

```python
# Grid untuk decision boundary
xx, yy = np.meshgrid(np.linspace(-0.5, 1.5, 200), np.linspace(-0.5, 1.5, 200))
grid = np.c_[xx.ravel(), yy.ravel()]

# Perceptron AND
p = Perceptron(2); p.latih(X, np.array([0,0,0,1]))
Z = np.array([p.prediksi(g) for g in grid]).reshape(xx.shape)
plt.contourf(xx, yy, Z, alpha=0.3)
plt.scatter(X[:,0], X[:,1], c=[0,0,0,1], s=100, edgecolors="k")
plt.title("Perceptron — AND (satu garis lurus)"); plt.show()
```

```python
# MLP XOR
Z = np.array([forward_mlp(g, XOR_W1, XOR_B1, XOR_W2, XOR_B2) for g in grid]).reshape(xx.shape)
plt.contourf(xx, yy, Z > 0.5, alpha=0.3)
plt.scatter(X[:,0], X[:,1], c=[0,1,1,0], s=100, edgecolors="k")
plt.title("MLP — XOR (wilayah non-linear)"); plt.show()
```
