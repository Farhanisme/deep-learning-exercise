# Eksplorasi Pertemuan 5 — Citra yang Salah Diklasifikasi
Tampilkan digit yang salah diprediksi — apakah kesalahannya masuk akal?

```python
import sys; sys.path.insert(0, "../src")
from digit import siapkan_digits, latih_mlp
X_train, X_test, y_train, y_test = siapkan_digits()
model = latih_mlp(X_train, y_train)
pred = model.predict(X_test)
```

```python
import matplotlib.pyplot as plt
import numpy as np
salah = np.where(pred != np.asarray(y_test))[0][:10]
fig, axes = plt.subplots(2, 5, figsize=(10, 4.5))
for ax, i in zip(axes.ravel(), salah):
    ax.imshow(X_test[i].reshape(8, 8), cmap="gray_r")
    ax.set_title(f"pred={pred[i]} asli={np.asarray(y_test)[i]}", fontsize=9)
    ax.axis("off")
plt.suptitle("Digit yang salah diklasifikasi"); plt.show()
```
